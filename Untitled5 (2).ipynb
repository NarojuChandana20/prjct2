```python
"""
Behaviour Tracker - Phase 1 (Rich Features Version)
---------------------------------------------------
Features per frame: 165
- 99  : Normalized landmarks (x, y, z)
- 33  : Joint distances
- 33  : Joint velocities

Input shape for model: (30, 165)
"""

from __future__ import annotations

import numpy as np
from pathlib import Path
from collections import deque
from typing import Dict, Optional, Tuple

import cv2
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
from mediapipe import Image as MPImage
import tensorflow as tf


CLASS_NAMES = {
    0: "NonFight / Normal",
    1: "Fight",
    2: "Fall",
}

RISK_LEVEL = {
    0: "Low",
    1: "High",
    2: "High",
}


def focal_loss(gamma=2.0, alpha=0.25):
    def loss(y_true, y_pred):
        y_pred = tf.clip_by_value(y_pred, 1e-7, 1.0 - 1e-7)
        cross_entropy = -y_true * tf.math.log(y_pred)
        weight = alpha * y_true * tf.pow(1 - y_pred, gamma)
        return tf.reduce_sum(weight * cross_entropy, axis=-1)
    return loss


class BehaviourTracker:
    def __init__(
        self,
        model_path: str | Path,
        pose_model_path: str | Path,
        sequence_length: int = 30,
        confidence_threshold: float = 0.55,
        smooth_window: int = 7,
    ):
        self.sequence_length = sequence_length
        self.confidence_threshold = confidence_threshold
        self.smooth_window = smooth_window

        # Load model
        model_path = Path(model_path)
        if not model_path.exists():
            raise FileNotFoundError(f"Behaviour model not found: {model_path}")

        self.model = tf.keras.models.load_model(
            str(model_path),
            custom_objects={"loss": focal_loss(gamma=2.0, alpha=0.25)}
        )
        print(f"[BehaviourTracker] Model loaded → {model_path.name}")

        # MediaPipe Pose
        base_options = python.BaseOptions(model_asset_path=str(pose_model_path))
        options = vision.PoseLandmarkerOptions(
            base_options=base_options,
            running_mode=vision.RunningMode.IMAGE,
            num_poses=1,
            min_pose_detection_confidence=0.5,
            min_pose_presence_confidence=0.5,
        )
        self.pose_landmarker = vision.PoseLandmarker.create_from_options(options)
        print("[BehaviourTracker] MediaPipe Pose Landmarker ready")

        # Buffers
        self.buffers: Dict[int, deque] = {}
        self.prev_coords: Dict[int, np.ndarray] = {}          # for velocity
        self.prediction_history: Dict[int, deque] = {}

    def _extract_features(self, frame: np.ndarray, track_id: int) -> Optional[np.ndarray]:
        """
        Returns 165 features or None
        """
        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        mp_image = MPImage(image_format=mp.ImageFormat.SRGB, data=rgb)
        result = self.pose_landmarker.detect(mp_image)

        if not result.pose_landmarks:
            return None

        landmarks = result.pose_landmarks[0]
        coords = np.array([[lm.x, lm.y, lm.z] for lm in landmarks], dtype=np.float32)  # (33, 3)

        # Quality check
        if np.mean(np.abs(coords[:, :2])) < 0.02:
            return None

        # ---------- Normalization ----------
        left_hip = coords[23]
        right_hip = coords[24]
        hip_center = (left_hip + right_hip) / 2.0
        coords = coords - hip_center

        left_shoulder = coords[11]
        right_shoulder = coords[12]
        shoulder_dist = np.linalg.norm(left_shoulder[:2] - right_shoulder[:2])
        if shoulder_dist < 1e-5:
            shoulder_dist = 1.0
        coords = coords / shoulder_dist

        # 1. Landmark positions (99)
        pos_features = coords.flatten()

        # 2. Joint distances (33)
        distances = []
        for i in range(32):
            dist = np.linalg.norm(coords[i] - coords[i + 1])
            distances.append(dist)
        distances.append(0.0)
        dist_features = np.array(distances, dtype=np.float32)

        # 3. Joint velocities (33)
        if track_id in self.prev_coords:
            velocity = np.linalg.norm(coords - self.prev_coords[track_id], axis=1)
        else:
            velocity = np.zeros(33, dtype=np.float32)

        # Save current coords for next frame
        self.prev_coords[track_id] = coords.copy()

        # Combine → 165 features
        features = np.concatenate([pos_features, dist_features, velocity])
        return features

    def update(self, track_id: int, person_crop: np.ndarray) -> Optional[Dict]:
        if person_crop is None or person_crop.size == 0:
            return None

        features = self._extract_features(person_crop, track_id)

        if track_id not in self.buffers:
            self.buffers[track_id] = deque(maxlen=self.sequence_length)
            self.prediction_history[track_id] = deque(maxlen=self.smooth_window)

        if features is None:
            if len(self.prediction_history[track_id]) > 0:
                return self._get_smoothed_prediction(track_id)
            return None

        self.buffers[track_id].append(features)

        if len(self.buffers[track_id]) < self.sequence_length:
            return None

        sequence = np.stack(list(self.buffers[track_id]), axis=0)  # (30, 165)
        raw_pred = self._predict(sequence)

        self.prediction_history[track_id].append(raw_pred["class_id"])
        return self._get_smoothed_prediction(track_id, raw_pred)

    def _predict(self, sequence: np.ndarray) -> Dict:
        x = np.expand_dims(sequence, axis=0).astype(np.float32)
        probs = self.model.predict(x, verbose=0)[0]

        class_id = int(np.argmax(probs))
        confidence = float(probs[class_id])

        if confidence < self.confidence_threshold and class_id != 0:
            class_id = 0
            confidence = float(probs[0])

        return {
            "class_id": class_id,
            "class_name": CLASS_NAMES[class_id],
            "risk_level": RISK_LEVEL[class_id],
            "confidence": round(confidence, 4),
            "probabilities": {
                "NonFight": round(float(probs[0]), 4),
                "Fight": round(float(probs[1]), 4),
                "Fall": round(float(probs[2]), 4),
            },
        }

    def _get_smoothed_prediction(self, track_id: int, raw_pred: Optional[Dict] = None) -> Dict:
        history = list(self.prediction_history[track_id])
        if not history:
            return raw_pred

        counts = np.bincount(history, minlength=3)
        smoothed_id = int(np.argmax(counts))

        if raw_pred is not None:
            result = raw_pred.copy()
            result["class_id"] = smoothed_id
            result["class_name"] = CLASS_NAMES[smoothed_id]
            result["risk_level"] = RISK_LEVEL[smoothed_id]
            return result
        else:
            return {
                "class_id": smoothed_id,
                "class_name": CLASS_NAMES[smoothed_id],
                "risk_level": RISK_LEVEL[smoothed_id],
                "confidence": 0.0,
                "probabilities": {"NonFight": 0.0, "Fight": 0.0, "Fall": 0.0},
            }

    def reset_track(self, track_id: int):
        if track_id in self.buffers:
            del self.buffers[track_id]
        if track_id in self.prediction_history:
            del self.prediction_history[track_id]
        if track_id in self.prev_coords:
            del self.prev_coords[track_id]

    def reset_all(self):
        self.buffers.clear()
        self.prediction_history.clear()
        self.prev_coords.clear()
```

```python
"""
Test BehaviourTracker - Save Output Video (Colab Compatible)
"""

import cv2
import numpy as np
from ultralytics import YOLO
from pathlib import Path
from google.colab.patches import cv2_imshow
import time

# ====================== PATHS ======================
VIDEO_PATH = "/content/drive/MyDrive/BehaviourRisk_System/models/WhatsApp Video 2026-09-19 at 14.25.42.mp4"
          # ← change this
MODEL_PATH = "/content/drive/MyDrive/BehaviourRisk_System/models/behaviour_model_rich_165_v2_best.keras"
POSE_MODEL_PATH = "/content/drive/MyDrive/BehaviourRisk_System/models/pose_landmarker_lite.task"

OUTPUT_PATH = "/content/drive/MyDrive/BehaviourRisk_System/models/output_behaviour_test.mp4"

# ====================== LOAD MODELS ======================
print("Loading models...")
yolo = YOLO("yolov8n.pt")

tracker = BehaviourTracker(
    model_path=MODEL_PATH,
    pose_model_path=POSE_MODEL_PATH,
    sequence_length=30,
    confidence_threshold=0.55,
    smooth_window=7
)
print("Models loaded successfully!\n")

# ====================== OPEN VIDEO ======================
cap = cv2.VideoCapture(VIDEO_PATH)
if not cap.isOpened():
    raise FileNotFoundError(f"Cannot open video: {VIDEO_PATH}")

# Get video properties
fps = int(cap.get(cv2.CAP_PROP_FPS))
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print(f"Video Info → {width}x{height} | FPS: {fps} | Total frames: {total_frames}")

# Create Video Writer
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(OUTPUT_PATH, fourcc, fps, (width, height))

# Simple tracker
next_id = 0
tracks = {}
MAX_DIST = 80

def get_track_id(cx, cy):
    global next_id
    best_id = None
    best_dist = MAX_DIST

    for tid, (px, py) in tracks.items():
        dist = np.sqrt((cx - px)**2 + (cy - py)**2)
        if dist < best_dist:
            best_dist = dist
            best_id = tid

    if best_id is not None:
        tracks[best_id] = (cx, cy)
        return best_id
    else:
        tracks[next_id] = (cx, cy)
        next_id += 1
        return next_id - 1

# ====================== PROCESS VIDEO ======================
print("Processing video... Please wait...")

frame_count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame_count += 1
    if frame_count % 30 == 0:
        print(f"Processed {frame_count}/{total_frames} frames...")

    results = yolo(frame, classes=[0], verbose=False)[0]

    for box in results.boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        conf = float(box.conf[0])

        if conf < 0.5:
            continue

        cx = (x1 + x2) // 2
        cy = (y1 + y2) // 2
        track_id = get_track_id(cx, cy)

        # Crop with padding
        h, w = frame.shape[:2]
        pad = 10
        x1c = max(0, x1 - pad)
        y1c = max(0, y1 - pad)
        x2c = min(w, x2 + pad)
        y2c = min(h, y2 + pad)

        person_crop = frame[y1c:y2c, x1c:x2c]

        # Get prediction
        prediction = tracker.update(track_id, person_crop)

        # Drawing
        color = (0, 255, 0)
        label = f"ID:{track_id} Collecting..."

        if prediction is not None:
            class_name = prediction["class_name"]
            conf = prediction["confidence"]
            risk = prediction["risk_level"]

            label = f"ID:{track_id} {class_name} ({conf:.2f})"
            color = (0, 0, 255) if risk == "High" else (0, 255, 0)

        cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
        cv2.putText(frame, label, (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.65, color, 2)

    # Write frame to output video
    out.write(frame)

# ====================== CLEANUP ======================
cap.release()
out.release()
tracker.reset_all()

print("\n✅ Processing finished!")
print(f"✅ Output video saved at:\n{OUTPUT_PATH}")
```

```python
"""
Train Improved Behaviour Model with Rich Features (165)
"""

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Bidirectional, LSTM, Dense, Dropout, BatchNormalization, GlobalAveragePooling1D
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.utils.class_weight import compute_class_weight
from sklearn.utils import resample
from collections import Counter
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns

# ====================== PATHS ======================
DATA_DIR = Path("/content/drive/MyDrive/BehaviourRisk_System/final_dataset_rich")
SAVE_DIR = Path("/content/drive/MyDrive/BehaviourRisk_System/models")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

# ====================== 1. Load Rich Data ======================
X = np.load(DATA_DIR / "final_X_rich.npy")
y = np.load(DATA_DIR / "final_y.npy")

print("Rich dataset shape:", X.shape)
print("Original distribution:", dict(Counter(y)))

# ====================== 2. Oversample Fall class ======================
X_0 = X[y == 0]
X_1 = X[y == 1]
X_2 = X[y == 2]

y_0 = y[y == 0]
y_1 = y[y == 1]
y_2 = y[y == 2]

target_fall = 2500

X_2_up, y_2_up = resample(X_2, y_2, replace=True, n_samples=target_fall, random_state=42)

X_balanced = np.concatenate([X_0, X_1, X_2_up], axis=0)
y_balanced = np.concatenate([y_0, y_1, y_2_up], axis=0)

print("\nAfter oversampling Fall:")
print("New distribution:", dict(Counter(y_balanced)))
print("New X shape:", X_balanced.shape)

# ====================== 3. Train / Val Split ======================
X_train, X_val, y_train, y_val = train_test_split(
    X_balanced, y_balanced,
    test_size=0.2,
    random_state=42,
    stratify=y_balanced
)

print(f"\nTrain: {X_train.shape} | Val: {X_val.shape}")

y_train_cat = to_categorical(y_train, num_classes=3)
y_val_cat   = to_categorical(y_val, num_classes=3)

# ====================== 4. Class Weights ======================
class_weights = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
class_weight_dict = dict(enumerate(class_weights))
print("Class weights:", class_weight_dict)

# ====================== 5. Focal Loss ======================
def focal_loss(gamma=2.0, alpha=0.25):
    def loss(y_true, y_pred):
        y_pred = tf.clip_by_value(y_pred, 1e-7, 1.0 - 1e-7)
        cross_entropy = -y_true * tf.math.log(y_pred)
        weight = alpha * y_true * tf.pow(1 - y_pred, gamma)
        return tf.reduce_sum(weight * cross_entropy, axis=-1)
    return loss

# ====================== 6. Build Stronger Model ======================
model = Sequential([
    Input(shape=(30, 165)),                          # ← Changed to 165

    Bidirectional(LSTM(128, return_sequences=True)),
    Dropout(0.3),

    Bidirectional(LSTM(64, return_sequences=True)),
    Dropout(0.3),

    Bidirectional(LSTM(32, return_sequences=True)),
    Dropout(0.25),

    GlobalAveragePooling1D(),

    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),

    Dense(32, activation='relu'),
    Dropout(0.2),

    Dense(3, activation='softmax')
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss=focal_loss(gamma=2.0, alpha=0.25),
    metrics=['accuracy']
)

model.summary()

# ====================== 7. Callbacks ======================
callbacks = [
    EarlyStopping(monitor='val_loss', patience=12, restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6, verbose=1),
    ModelCheckpoint(
        filepath=str(SAVE_DIR / "behaviour_model_rich_165_best.keras"),
        monitor='val_accuracy',
        save_best_only=True,
        verbose=1
    )
]

# ====================== 8. Train ======================
history = model.fit(
    X_train, y_train_cat,
    validation_data=(X_val, y_val_cat),
    epochs=50,
    batch_size=32,
    class_weight=class_weight_dict,
    callbacks=callbacks,
    verbose=1
)

# ====================== 9. Evaluate ======================
print("\n===== Final Evaluation =====")
val_loss, val_acc = model.evaluate(X_val, y_val_cat, verbose=0)
print(f"Val Loss: {val_loss:.4f}  |  Val Accuracy: {val_acc:.4f}")

y_pred = np.argmax(model.predict(X_val, verbose=0), axis=1)
y_true = y_val

print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=['NonFight', 'Fight', 'Fall']))

print("\nConfusion Matrix:")
cm = confusion_matrix(y_true, y_pred)
print(cm)

plt.figure(figsize=(7, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['NonFight', 'Fight', 'Fall'],
            yticklabels=['NonFight', 'Fight', 'Fall'])
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Rich Features (165) - Confusion Matrix')
plt.show()

# ====================== 10. Save Final Model ======================
final_path = SAVE_DIR / "behaviour_model_rich_165_final.keras"
model.save(final_path)

print(f"\n✅ Best model saved to: {SAVE_DIR / 'behaviour_model_rich_165_best.keras'}")
print(f"✅ Final model also saved to: {final_path}")
```

```python
"""
Train Improved Behaviour Model (Option 1)
- Rich features (165)
- Higher weight for NonFight class
"""

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Bidirectional, LSTM, Dense, Dropout, BatchNormalization, GlobalAveragePooling1D
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.utils import resample
from collections import Counter
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns

# ====================== PATHS ======================
DATA_DIR = Path("/content/drive/MyDrive/BehaviourRisk_System/final_dataset_rich")
SAVE_DIR = Path("/content/drive/MyDrive/BehaviourRisk_System/models")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

# ====================== 1. Load Data ======================
X = np.load(DATA_DIR / "final_X_rich.npy")
y = np.load(DATA_DIR / "final_y.npy")

print("Dataset shape:", X.shape)
print("Original distribution:", dict(Counter(y)))

# ====================== 2. Oversample Fall ======================
X_0 = X[y == 0]
X_1 = X[y == 1]
X_2 = X[y == 2]

y_0 = y[y == 0]
y_1 = y[y == 1]
y_2 = y[y == 2]

target_fall = 2500
X_2_up, y_2_up = resample(X_2, y_2, replace=True, n_samples=target_fall, random_state=42)

X_balanced = np.concatenate([X_0, X_1, X_2_up], axis=0)
y_balanced = np.concatenate([y_0, y_1, y_2_up], axis=0)

print("\nAfter oversampling Fall:")
print("New distribution:", dict(Counter(y_balanced)))

# ====================== 3. Train / Val Split ======================
X_train, X_val, y_train, y_val = train_test_split(
    X_balanced, y_balanced, test_size=0.2, random_state=42, stratify=y_balanced
)

y_train_cat = to_categorical(y_train, num_classes=3)
y_val_cat   = to_categorical(y_val, num_classes=3)

print(f"Train: {X_train.shape} | Val: {X_val.shape}")

# ====================== 4. Custom Class Weights ======================
# Higher weight for NonFight (class 0) to reduce false Fight predictions
class_weight_dict = {
    0: 1.8,    # NonFight  ← increased
    1: 1.0,    # Fight
    2: 1.2     # Fall
}
print("Using class weights:", class_weight_dict)

# ====================== 5. Focal Loss ======================
def focal_loss(gamma=2.0, alpha=0.25):
    def loss(y_true, y_pred):
        y_pred = tf.clip_by_value(y_pred, 1e-7, 1.0 - 1e-7)
        cross_entropy = -y_true * tf.math.log(y_pred)
        weight = alpha * y_true * tf.pow(1 - y_pred, gamma)
        return tf.reduce_sum(weight * cross_entropy, axis=-1)
    return loss

# ====================== 6. Build Model ======================
model = Sequential([
    Input(shape=(30, 165)),

    Bidirectional(LSTM(128, return_sequences=True)),
    Dropout(0.35),

    Bidirectional(LSTM(64, return_sequences=True)),
    Dropout(0.35),

    Bidirectional(LSTM(32, return_sequences=True)),
    Dropout(0.3),

    GlobalAveragePooling1D(),

    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(0.35),

    Dense(32, activation='relu'),
    Dropout(0.25),

    Dense(3, activation='softmax')
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=7e-4),   # slightly lower LR
    loss=focal_loss(gamma=2.0, alpha=0.25),
    metrics=['accuracy']
)

model.summary()

# ====================== 7. Callbacks ======================
callbacks = [
    EarlyStopping(monitor='val_loss', patience=12, restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6, verbose=1),
    ModelCheckpoint(
        filepath=str(SAVE_DIR / "behaviour_model_rich_165_v2_best.keras"),
        monitor='val_accuracy',
        save_best_only=True,
        verbose=1
    )
]

# ====================== 8. Train ======================
history = model.fit(
    X_train, y_train_cat,
    validation_data=(X_val, y_val_cat),
    epochs=50,
    batch_size=32,
    class_weight=class_weight_dict,
    callbacks=callbacks,
    verbose=1
)

# ====================== 9. Evaluate ======================
print("\n===== Final Evaluation =====")
val_loss, val_acc = model.evaluate(X_val, y_val_cat, verbose=0)
print(f"Val Loss: {val_loss:.4f}  |  Val Accuracy: {val_acc:.4f}")

y_pred = np.argmax(model.predict(X_val, verbose=0), axis=1)

print("\nClassification Report:")
print(classification_report(y_val, y_pred, target_names=['NonFight', 'Fight', 'Fall']))

print("\nConfusion Matrix:")
cm = confusion_matrix(y_val, y_pred)
print(cm)

plt.figure(figsize=(7, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['NonFight', 'Fight', 'Fall'],
            yticklabels=['NonFight', 'Fight', 'Fall'])
plt.title('Rich Features v2 - Confusion Matrix')
plt.show()

# ====================== 10. Save ======================
final_path = SAVE_DIR / "behaviour_model_rich_165_v2_final.keras"
model.save(final_path)

print(f"\n✅ Best model: {SAVE_DIR / 'behaviour_model_rich_165_v2_best.keras'}")
print(f"✅ Final model: {final_path}")
```

```python
"""
Behaviour Tracker - Phase 1 (Rich Features Version) - Improved
-------------------------------------------------------------
Features per frame: 165
- 99  : Normalized landmarks (x, y, z)
- 33  : Joint distances
- 33  : Joint velocities

Input shape for model: (30, 165)
"""

from __future__ import annotations

import numpy as np
from pathlib import Path
from collections import deque
from typing import Dict, Optional, Tuple

import cv2
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
from mediapipe import Image as MPImage
import tensorflow as tf


CLASS_NAMES = {
    0: "NonFight / Normal",
    1: "Fight",
    2: "Fall",
}

RISK_LEVEL = {
    0: "Low",
    1: "High",
    2: "High",
}


def focal_loss(gamma=2.0, alpha=0.25):
    def loss(y_true, y_pred):
        y_pred = tf.clip_by_value(y_pred, 1e-7, 1.0 - 1e-7)
        cross_entropy = -y_true * tf.math.log(y_pred)
        weight = alpha * y_true * tf.pow(1 - y_pred, gamma)
        return tf.reduce_sum(weight * cross_entropy, axis=-1)
    return loss


class BehaviourTracker:
    def __init__(
        self,
        model_path: str | Path,
        pose_model_path: str | Path,
        sequence_length: int = 30,
        confidence_threshold: float = 0.55,
        smooth_window: int = 7,
        max_fail_frames: int = 12,          # how many consecutive bad frames before clearing buffer
    ):
        self.sequence_length = sequence_length
        self.confidence_threshold = confidence_threshold
        self.smooth_window = smooth_window
        self.max_fail_frames = max_fail_frames

        # Load model
        model_path = Path(model_path)
        if not model_path.exists():
            raise FileNotFoundError(f"Behaviour model not found: {model_path}")

        self.model = tf.keras.models.load_model(
            str(model_path),
            custom_objects={"loss": focal_loss(gamma=2.0, alpha=0.25)}
        )
        print(f"[BehaviourTracker] Model loaded → {model_path.name}")

        # MediaPipe Pose
        base_options = python.BaseOptions(model_asset_path=str(pose_model_path))
        options = vision.PoseLandmarkerOptions(
            base_options=base_options,
            running_mode=vision.RunningMode.IMAGE,
            num_poses=1,
            min_pose_detection_confidence=0.45,   # slightly lower for better recall during fights
            min_pose_presence_confidence=0.45,
        )
        self.pose_landmarker = vision.PoseLandmarker.create_from_options(options)
        print("[BehaviourTracker] MediaPipe Pose Landmarker ready")

        # Buffers
        self.buffers: Dict[int, deque] = {}
        self.prev_coords: Dict[int, np.ndarray] = {}
        self.prediction_history: Dict[int, deque] = {}
        self.fail_count: Dict[int, int] = {}
        self.last_good_pred: Dict[int, Dict] = {}   # keep last good result for a few frames

    def _extract_features(self, frame: np.ndarray, track_id: int) -> Optional[np.ndarray]:
        """
        Returns 165 features or None
        """
        if frame is None or frame.size == 0:
            return None

        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        mp_image = MPImage(image_format=mp.ImageFormat.SRGB, data=rgb)
        result = self.pose_landmarker.detect(mp_image)

        if not result.pose_landmarks:
            return None

        landmarks = result.pose_landmarks[0]
        coords = np.array([[lm.x, lm.y, lm.z] for lm in landmarks], dtype=np.float32)  # (33, 3)

        # ---------- Quality check (much more tolerant) ----------
        if np.mean(np.abs(coords[:, :2])) < 0.005:          # was 0.02 → now 0.005
            return None

        # ---------- Normalization ----------
        left_hip = coords[23]
        right_hip = coords[24]
        hip_center = (left_hip + right_hip) / 2.0
        coords = coords - hip_center

        left_shoulder = coords[11]
        right_shoulder = coords[12]
        shoulder_dist = np.linalg.norm(left_shoulder[:2] - right_shoulder[:2])
        if shoulder_dist < 1e-5:
            shoulder_dist = 1.0
        coords = coords / shoulder_dist

        # 1. Landmark positions (99)
        pos_features = coords.flatten()

        # 2. Joint distances (33)
        distances = []
        for i in range(32):
            dist = np.linalg.norm(coords[i] - coords[i + 1])
            distances.append(dist)
        distances.append(0.0)  # pad to 33
        dist_features = np.array(distances, dtype=np.float32)

        # 3. Joint velocities (33)
        if track_id in self.prev_coords:
            velocity = np.linalg.norm(coords - self.prev_coords[track_id], axis=1)
        else:
            velocity = np.zeros(33, dtype=np.float32)

        self.prev_coords[track_id] = coords.copy()

        # Combine → 165 features
        features = np.concatenate([pos_features, dist_features, velocity])
        return features

    def update(self, track_id: int, person_crop: np.ndarray) -> Optional[Dict]:
        if person_crop is None or person_crop.size == 0:
            return self._handle_failed_frame(track_id)

        features = self._extract_features(person_crop, track_id)

        # Initialize buffers if new track
        if track_id not in self.buffers:
            self.buffers[track_id] = deque(maxlen=self.sequence_length)
            self.prediction_history[track_id] = deque(maxlen=self.smooth_window)
            self.fail_count[track_id] = 0

        if features is None:
            return self._handle_failed_frame(track_id)

        # Good frame → reset fail counter and add features
        self.fail_count[track_id] = 0
        self.buffers[track_id].append(features)

        # Still collecting
        if len(self.buffers[track_id]) < self.sequence_length:
            collected = len(self.buffers[track_id])
            return {
                "status": "collecting",
                "class_id": -1,
                "class_name": f"Collecting {collected}/{self.sequence_length}",
                "risk_level": "—",
                "confidence": 0.0,
                "probabilities": {"NonFight": 0.0, "Fight": 0.0, "Fall": 0.0},
                "collected": collected,
            }

        # Ready for prediction
        sequence = np.stack(list(self.buffers[track_id]), axis=0)  # (30, 165)
        raw_pred = self._predict(sequence)

        self.prediction_history[track_id].append(raw_pred["class_id"])
        smoothed = self._get_smoothed_prediction(track_id, raw_pred)

        # Save as last good prediction
        self.last_good_pred[track_id] = smoothed
        return smoothed

    def _handle_failed_frame(self, track_id: int) -> Optional[Dict]:
        """Called when pose detection fails"""
        if track_id not in self.buffers:
            return None

        self.fail_count[track_id] = self.fail_count.get(track_id, 0) + 1

        # Too many consecutive failures → clear buffer
        if self.fail_count[track_id] > self.max_fail_frames:
            self.buffers[track_id].clear()
            self.fail_count[track_id] = 0
            return {
                "status": "collecting",
                "class_id": -1,
                "class_name": f"Collecting 0/{self.sequence_length}",
                "risk_level": "—",
                "confidence": 0.0,
                "probabilities": {"NonFight": 0.0, "Fight": 0.0, "Fall": 0.0},
                "collected": 0,
            }

        # Still have some frames → show progress
        collected = len(self.buffers[track_id])
        if collected < self.sequence_length:
            return {
                "status": "collecting",
                "class_id": -1,
                "class_name": f"Collecting {collected}/{self.sequence_length}",
                "risk_level": "—",
                "confidence": 0.0,
                "probabilities": {"NonFight": 0.0, "Fight": 0.0, "Fall": 0.0},
                "collected": collected,
            }

        # We already have a full sequence → return last good prediction for a while
        if track_id in self.last_good_pred:
            return self.last_good_pred[track_id]

        return None

    def _predict(self, sequence: np.ndarray) -> Dict:
        x = np.expand_dims(sequence, axis=0).astype(np.float32)
        probs = self.model.predict(x, verbose=0)[0]

        class_id = int(np.argmax(probs))
        confidence = float(probs[class_id])

        # Force NonFight if confidence is low (except for Fall which is usually high conf)
        if confidence < self.confidence_threshold and class_id != 2:
            class_id = 0
            confidence = float(probs[0])

        return {
            "class_id": class_id,
            "class_name": CLASS_NAMES[class_id],
            "risk_level": RISK_LEVEL[class_id],
            "confidence": round(confidence, 4),
            "probabilities": {
                "NonFight": round(float(probs[0]), 4),
                "Fight": round(float(probs[1]), 4),
                "Fall": round(float(probs[2]), 4),
            },
        }

    def _get_smoothed_prediction(self, track_id: int, raw_pred: Optional[Dict] = None) -> Dict:
        history = list(self.prediction_history.get(track_id, []))
        if not history:
            return raw_pred

        counts = np.bincount(history, minlength=3)
        smoothed_id = int(np.argmax(counts))

        if raw_pred is not None:
            result = raw_pred.copy()
            result["class_id"] = smoothed_id
            result["class_name"] = CLASS_NAMES[smoothed_id]
            result["risk_level"] = RISK_LEVEL[smoothed_id]
            return result

        return {
            "class_id": smoothed_id,
            "class_name": CLASS_NAMES[smoothed_id],
            "risk_level": RISK_LEVEL[smoothed_id],
            "confidence": 0.0,
            "probabilities": {"NonFight": 0.0, "Fight": 0.0, "Fall": 0.0},
        }

    def reset_track(self, track_id: int):
        for d in [self.buffers, self.prediction_history, self.prev_coords, self.fail_count, self.last_good_pred]:
            if track_id in d:
                del d[track_id]

    def reset_all(self):
        self.buffers.clear()
        self.prediction_history.clear()
        self.prev_coords.clear()
        self.fail_count.clear()
        self.last_good_pred.clear()
```

```python
"""
Test BehaviourTracker on a video (Colab version)
"""

import cv2
from pathlib import Path
from ultralytics import YOLO
#from behaviour_tracker import BehaviourTracker
from google.colab.patches import cv2_imshow   # for displaying in Colab
from IPython.display import clear_output
import time

# ====================== CHANGE THESE PATHS ======================
VIDEO_PATH = "/content/drive/MyDrive/BehaviourRisk_System/models/WhatsApp Video 2026-09-19 at 14.25.42.mp4"
OUTPUT_PATH = "/content/drive/MyDrive/result_output.mp4"      # ← where to save result

MODEL_PATH = " /content/drive/MyDrive/BehaviourRisk_System/models/behaviour_model_rich_165_v2_best.keras"
POSE_MODEL_PATH =" /content/drive/MyDrive/BehaviourRisk_System/models/pose_landmarker_lite.task"


# ================================================================

print("[INFO] Loading YOLO...")
yolo = YOLO("yolo11n.pt")

print("[INFO] Loading BehaviourTracker...")
tracker = BehaviourTracker(
    model_path=MODEL_PATH,
    pose_model_path=POSE_MODEL_PATH,
    sequence_length=30,
    confidence_threshold=0.55,
    smooth_window=7,
    max_fail_frames=12,
)

cap = cv2.VideoCapture(VIDEO_PATH)
if not cap.isOpened():
    raise FileNotFoundError(f"Cannot open video: {VIDEO_PATH}")

width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps    = cap.get(cv2.CAP_PROP_FPS) or 25
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(OUTPUT_PATH, fourcc, fps, (width, height))

print(f"[INFO] Processing → {VIDEO_PATH}")
print(f"[INFO] Total frames: {total_frames}")
print(f"[INFO] Output    → {OUTPUT_PATH}")
print("-" * 60)

frame_idx = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame_idx += 1

    results = yolo.track(
        frame,
        persist=True,
        classes=[0],
        conf=0.4,
        iou=0.5,
        verbose=False,
    )

    if results[0].boxes.id is not None:
        boxes = results[0].boxes.xyxy.cpu().numpy()
        track_ids = results[0].boxes.id.int().cpu().tolist()

        for box, track_id in zip(boxes, track_ids):
            x1, y1, x2, y2 = map(int, box)
            x1, y1 = max(0, x1), max(0, y1)
            x2, y2 = min(width - 1, x2), min(height - 1, y2)

            person_crop = frame[y1:y2, x1:x2]
            result = tracker.update(track_id, person_crop)

            if result is None:
                label = f"ID:{track_id} Lost"
                color = (128, 128, 128)
            elif result.get("status") == "collecting":
                label = f"ID:{track_id} {result['class_name']}"
                color = (0, 255, 255)          # Yellow
            else:
                label = f"ID:{track_id} {result['class_name']} {result['confidence']:.2f}"
                if result["class_id"] == 1:    # Fight
                    color = (0, 0, 255)
                elif result["class_id"] == 2:  # Fall
                    color = (0, 165, 255)
                else:
                    color = (0, 255, 0)

            cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
            (w, h), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.65, 2)
            cv2.rectangle(frame, (x1, y1 - 28), (x1 + w + 8, y1), color, -1)
            cv2.putText(frame, label, (x1 + 4, y1 - 8),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.65, (255, 255, 255), 2)

    cv2.putText(frame, f"Frame: {frame_idx}/{total_frames}", (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)

    out.write(frame)

    # Show every 5th frame in Colab (to avoid slowing down too much)
    if frame_idx % 5 == 0:
        clear_output(wait=True)
        cv2_imshow(frame)
        print(f"Processed {frame_idx}/{total_frames} frames...")

cap.release()
out.release()
tracker.reset_all()

print("-" * 60)
print(f"[DONE] Saved → {OUTPUT_PATH}")
```
